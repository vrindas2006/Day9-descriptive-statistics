# **DAY 9 – DESCRIPTIVE STATISTICS**





## Introduction

Descriptive statistics are used to summarize and understand the main characteristics of a dataset.

In this task, the Titanic dataset is analyzed using descriptive statistical measures such as:

- Mean
- Median
- Mode
- Standard Deviation
- Percentiles
- Skewness

The analysis focuses on numerical columns to understand their central tendency, variability, and distribution.

## Objective

The objective of this task is to calculate and interpret descriptive statistics for the key numerical columns of the Titanic dataset.

The analysis aims to:

1. Identify the average and typical values using mean and median.
2. Find the most frequently occurring values using mode.
3. Measure the spread of data using standard deviation.
4. Understand the distribution using percentiles.
5. Compare mean and median to identify possible skewness.
6. Understand what these statistics reveal about the dataset.

```python
# Importing Libraries
import pandas as pd

# Read titanic dataset
df=pd.read_csv("/Titanic-Dataset.csv")
df
```

```python
# View first 5 row and colums
display(df.head())

# Print Columns In Dataset
print(df.columns)

# Check Dataset Information
print(df.info())
print(df.shape)
df.isnull().sum()
```

```python
# Select Numeric Columns
numeric_cols=["Pclass","Age","SibSp","Parch","Fare"]
df[numeric_cols].describe()
```

# Calculate Mean

### Objective
To calculate the mean of each numerical variable and understand the average value in the dataset.

### Procedure
The mean is calculated by adding all the values and dividing the total by the number of observations.

Mean helps identify the average or central value of a numerical variable.

```python
# Calculating Mean of Numeric Columns
df[numeric_cols].mean()
```

### Finding

The mean represents the average value of each numerical variable.

The average passenger age is approximately 30years, while the average fare is approximately 33.

The average values of SibSp and Parch indicate the typical number of family members related to passengers aboard the Titanic.

The mean provides a useful measure of central tendency, but it can be influenced by extremely high or low values.

# **Calculate Median**

### Objective
To determine the middle value of each numerical variable after arranging the observations in ascending order.

### Procedure
The median divides the dataset into two equal parts:
- 50% of observations are below the median.
- 50% of observations are above the median.

Median is particularly useful when the data contains extreme values or is skewed.

```python
# Calculate Median of Numeric Columns
df[numeric_cols].median()
```

### Finding

The median age of the passengers is 28 years, meaning that 50% of the passengers are aged below 28 years and 50% are above 28 years.

The median Pclass is 3, indicating that the middle observation belongs to the third passenger class.

The median values of SibSp and Parch are both 0, showing that at least half of the passengers had no siblings/spouses or parents/children aboard.

The median Fare is approximately 14.45, meaning that half of the passengers paid a fare below approximately 14.45 and half paid more.

The median is useful for understanding the typical value because it is less affected by extreme values than the mean.

# **Compare Mean vs Median**

```python
comparison=pd.DataFrame({"Mean":df[numeric_cols].mean(),"Median":df[numeric_cols].median()})
comparison.round(2)
```

### Finding

The mean and median values were compared to identify possible skewness in the numerical variables.

- Pclass has a mean of 2.31 and a median of 3.00. Since the mean is lower than the median, the distribution shows a tendency toward left skewness.
- Age has a mean of 29.70 and a median of 28.00. Since the mean is slightly higher than the median, Age shows slight right skewness.
- SibSp has a mean of 0.52 and a median of 0.00, indicating right skewness.
- Parch has a mean of 0.38 and a median of 0.00, also indicating right skewness.
- Fare has a mean of 32.20 and a median of 14.45. The large difference between the mean and median indicates strong positive/right skewness. This is likely because a small number of passengers paid very high fares.

Overall, Fare shows the most noticeable right skewness among the selected variables.

# **Calculate Mode**

```python
df[numeric_cols].mode().iloc[0]
```

### Finding

The mode analysis shows the most frequently occurring value in each numerical variable.

Pclass has a mode of 3, indicating that third-class passengers were the most frequently occurring group.

The mode of Age is 24 years, meaning 24 is the most frequently occurring age in the dataset.

Both SibSp and Parch have a mode of 0, indicating that the most common passenger had no siblings/spouses or parents/children aboard.

The mode of Fare is 8.05, showing that this was the most frequently occurring ticket fare.

Overall, the mode helps identify the most common observations in the dataset.

#Calculate Standard Deviation

###Objectives
The objective is to measure how much the values in each numerical column vary or spread around their mean

```python
# Calculating Standard Deviation
df[numeric_cols].std()
```

### Finding

The standard deviation values show how much the observations vary around their respective mean values.

- Pclass has a standard deviation of 0.836, indicating relatively low variation in passenger class.
- Age has a standard deviation of 14.526, showing that passenger ages are spread noticeably around the mean age.
- SibSp has a standard deviation of 1.103, indicating some variation in the number of siblings or spouses aboard.
- Parch has a standard deviation of 0.806, indicating relatively low variation in the number of parents or children aboard.
- Fare has the highest standard deviation of 49.693, indicating a large variation in ticket fares among passengers.

Overall, Fare has the greatest variability among the selected variables, while Parch and Pclass show relatively smaller numerical variation.

#Fare is highly variable and strongly right-skewed

#Percentile Analysis
#Objective

The objective is to understand how the numerical values are distributed and identify the values below which 25%, 50%, and 75% of the observations fall.

The three important percentiles are:

25th percentile (Q1) → 25% of values are below this point.

50th percentile (Q2) → 50% of values are below this point. This is the median.

75th percentile (Q3) → 75% of values are below this point.

```python
#Calculate Percentile
df[numeric_cols].quantile([0.25,0.5,0.75])
```

### Finding

The percentile analysis provides information about the distribution of the selected numerical variables.

For Age, the 25th percentile is 20.125 years, the median is 28 years, and the 75th percentile is 38 years. Therefore, the middle 50% of passenger ages lie between 20.125 and 38 years.

For Fare, the 25th percentile is 7.9104, the median is 14.4542, and the 75th percentile is 31.00. Therefore, the middle 50% of ticket fares lie between 7.9104 and 31.00.

For SibSp, the 25th and 50th percentiles are 0, while the 75th percentile is 1, showing that most passengers had very few siblings or spouses aboard.

For Parch, the 25th, 50th, and 75th percentiles are all 0, indicating that at least 75% of the observations have no parents or children aboard.

The percentile values, particularly for Fare, show that most observations are concentrated at lower values, while a smaller number of passengers have much higher fares.

#Calculate InterQuartile Range(IQR)

Now we need to calculate the Interquartile Range (IQR) to specifically describe the spread of the middle 50%, which is part of the percentile analysis.

```python
#Calculate interquartile range
df[numeric_cols].quantile(0.75)-df[numeric_cols].quantile(0.25)
```

### Finding

The IQR values show the spread of the middle 50% of observations.

For Age, the IQR is 17.875 years, meaning that the middle 50% of passenger ages are spread across a range of 17.875 years, from 20.125 to 38 years.

For Fare, the IQR is 23.0896, meaning that the middle 50% of ticket fares range from 7.9104 to 31.00.

SibSp has an IQR of 1, indicating that the middle 50% of observations lie between 0 and 1 sibling/spouse.

Parch has an IQR of 0 because both its 25th and 75th percentiles are 0, showing that the middle 50% of observations are concentrated at 0.

Pclass has an IQR of 1, showing that the middle 50% of observations span one passenger-class level.

Overall, Fare and Age show greater spread in their middle 50% compared with Parch and the other variables.

#Skewness Analysis
### Objective

The objective of skewness analysis is to determine the degree and direction of asymmetry in the distribution of the numerical variables.

Skewness helps identify whether the data is:

- Positively skewed (right-skewed)
- Negatively skewed (left-skewed)
- Approximately symmetric

By analyzing skewness, we can understand how the observations are distributed around the mean.

```python
#Skewness
df[numeric_cols].skew()
```

### Finding

The skewness analysis shows that the selected numerical variables have different distribution patterns.

- Pclass has a skewness of -0.631, indicating a negative or left-skewed distribution.
- Age has a skewness of 0.389, indicating slight positive or right skewness. The distribution of Age is relatively close to being symmetric.
- SibSp has a skewness of 3.695, indicating strong positive or right skewness. Most passengers had few or no siblings/spouses, while a small number had several.
- Parch has a skewness of 2.749, indicating strong positive or right skewness. Most passengers had no parents or children aboard, while a smaller number had larger family groups.
- Fare has a skewness of 4.787, indicating very strong positive or right skewness. Most passengers paid relatively lower fares, while a small number paid very high fares.

Overall, Fare has the highest positive skewness among the selected variables, followed by SibSp and Parch. Age is relatively close to symmetric compared with the other variables.

#Create a Summary Statistics Table
Objective

The objective is to create a summary table containing the main descriptive statistics for all selected numerical variables.

The table will include:

Mean

Median

Mode

Standard Deviation

25th Percentile

50th Percentile

75th Percentile

Skewness

```python
summary = pd.DataFrame({
    'Mean': df[numeric_cols].mean(),
    'Median': df[numeric_cols].median(),
    'Mode': df[numeric_cols].mode().iloc[0],
    'Standard Deviation': df[numeric_cols].std(),
    '25th Percentile': df[numeric_cols].quantile(0.25),
    '50th Percentile': df[numeric_cols].quantile(0.50),
    '75th Percentile': df[numeric_cols].quantile(0.75),
    'Skewness': df[numeric_cols].skew()
})

summary.round(2)
```

### Finding

The summary statistics table combines the major descriptive measures for the selected numerical variables.

The results show that Age has a mean of 29.70 years and a median of 28 years, indicating a relatively small difference between the two measures.

Fare has a mean of 32.20 and a median of 14.45, along with a high standard deviation of 49.69 and skewness of 4.79. This indicates high variability and strong positive/right skewness in the Fare variable.

SibSp and Parch have median and mode values of 0, showing that having no siblings/spouses or parents/children aboard was common among the passengers.

Overall, the summary table provides a compact overview of the central tendency, variability, percentiles, and distribution of the selected variables.

#Create a Histogram

Now we'll create a histogram for the Fare variable. This gives you a visual representation of the distribution and helps confirm the skewness we calculated earlier.

###Objective

The objective is to visualize the distribution of passenger fares and identify whether the data is symmetric or skewed.

###Procedure

We will use Matplotlib to create a histogram of the Fare colum

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.hist(df[numeric_cols]['Fare'], bins=30)

plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.title('Distribution of Passenger Fare')

plt.show()
```

### Finding

The histogram shows that most passengers paid relatively lower fares, while a smaller number of passengers paid substantially higher fares.

The distribution has a long tail toward the higher fare values, indicating that the Fare variable is positively or right-skewed.

This observation is consistent with the skewness value of 4.79 and the difference between the mean (32.20) and median (14.45).

# Conclusion

The descriptive statistical analysis provided useful insights into the central tendency, variability, and distribution of the numerical variables in the Titanic dataset.

Mean, median, and mode were used to understand the typical values of the variables. Standard deviation and IQR helped measure the spread and variability of the data, while percentiles provided information about the distribution of the observations.

The analysis showed that Fare has high variability and strong positive/right skewness. The histogram also confirmed that most passengers paid lower fares, while a smaller number paid much higher fares.

Overall, descriptive statistics provide an effective way to understand the characteristics and distribution of numerical data before performing further analysis.
